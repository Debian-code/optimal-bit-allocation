```python
from numerical_method import NumericalMethod
import numpy as np
import matplotlib.pyplot as plt
```

```python
sigmas = np.sqrt([0.00255, 0.00781, 0.01562])

m = NumericalMethod(B=12, sigmas=sigmas)
```

```python
b_final, history = m.projected_gradient_descent(b0=[0, 0, 12], learning_rate=50.0)
```

```python
b_star = m.get_solution()
print("b* =", b_star, "   b found =", b_final)
```

```python
fig = m.display_followed_path(history, b_star=b_star, title="Projected gradient, B = 12")
plt.show()
```

```python
b, mu, lam, history = m.uzawa(rho=1e-5)
print(m.lagrangian_derivative(b, mu, lam))   # should be ≈ (0, 0, 0)
fig = m.display_followed_path(history, title="Uzawa")
```
